# Model Evaluation and Analysis

Comprehensive evaluation of trained segmentation models.

```python
import torch
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import json

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (15, 10)
```

## Load Evaluation Results

```python
# Load evaluation results
results_dir = Path('results')
eval_files = list(results_dir.glob('evaluation_*.json'))

if eval_files:
    latest_eval = sorted(eval_files)[-1]
    with open(latest_eval, 'r') as f:
        results = json.load(f)
    
    print('Evaluation Results:')
    for metric, stats in results.items():
        print(f'\n{metric.upper()}:')
        for key, value in stats.items():
            print(f'  {key}: {value:.4f}')
else:
    print('No evaluation results found')
```

## Metrics Visualization

```python
# Visualize metrics
if 'results' in locals():
    metrics = list(results.keys())
    means = [results[m]['mean'] for m in metrics]
    stds = [results[m]['std'] for m in metrics]
    
    fig, ax = plt.subplots(figsize=(12, 6))
    ax.bar(metrics, means, yerr=stds, capsize=10, color='steelblue', alpha=0.7)
    ax.set_ylabel('Score')
    ax.set_title('Segmentation Metrics (Mean ± Std)')
    ax.grid(axis='y', alpha=0.3)
    plt.tight_layout()
    plt.show()
```

## View Predictions

```python
# Display prediction visualizations
pred_dir = Path('results')
pred_images = sorted(pred_dir.glob('prediction_*.png'))

if pred_images:
    fig, axes = plt.subplots(len(pred_images[:4]), 1, figsize=(15, 5*len(pred_images[:4])))
    
    for idx, img_path in enumerate(pred_images[:4]):
        from PIL import Image
        img = Image.open(img_path)
        axes[idx].imshow(img)
        axes[idx].set_title(f'Prediction {idx}')
        axes[idx].axis('off')
    
    plt.tight_layout()
    plt.show()
else:
    print('No prediction visualizations found')
```
