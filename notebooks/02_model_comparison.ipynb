# Model Comparison Analysis

Compare performance metrics of different segmentation models.

```python
import torch
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import json

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (15, 10)
```

## Run Model Comparison

```python
from scripts.compare_models import ModelComparator

# Run comparison
comparator = ModelComparator(device='cuda' if torch.cuda.is_available() else 'cpu')
results = comparator.compare_models()
comparator.print_summary()
```

## Visualize Results

```python
comparator.visualize_comparison()
comparator.save_comparison_report()
```

## Detailed Analysis

```python
# Extract data for detailed analysis
models = list(results.keys())
model_names = [results[m]['name'] for m in models]

# Create detailed comparison table
print('\n' + '='*80)
print('DETAILED MODEL COMPARISON')
print('='*80 + '\n')

print(f'{"Model":<25} {"Type":<15} {"Size (MB)":<12} {"Speed (ms)":<12} {"Memory (MB)":<12}')
print('-' * 80)

for model in models:
    r = results[model]
    print(f"{r['name']:<25} {r['type']:<15} {r['model_size_mb']:<12.2f} {r['inference_time_ms']:<12.2f} {r['peak_memory_mb']:<12.2f}")

print('\n' + '='*80)
```
